```python
with open("gym_data.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

repeated = []
one_time = []

for row in rows:
    Member_ID = row["Member_ID"]
    if Member_ID in one_time:
        if Member_ID not in repeated:
            repeated.append(Member_ID)
    else:
        one_time.append(Member_ID)

print(repeated)
```
